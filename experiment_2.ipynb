```python
import numpy as np
import pandas as pd

df = pd.read_csv("Rajasthan_Heatwave_2006_2025.csv")
```

```python
print(df.head())
```

```python
print(df.info())
```

```python
print(df.describe())
```

```python
print("Shape:", df.shape)
print("Columns:", df.columns)
```

```python
print(df.isnull().sum())
```

```python
print("Total null values:", df.isnull().sum().sum())
```

```python
df["TEMP2M"].fillna(df["TEMP2M"].mean(), inplace=True)
```

```python
print(df.isnull().sum())
```

```python
df["TEMP2M"] = df["TEMP2M"].fillna(df["TEMP2M"].mean())
```

```python
print(df[df["HEATWAVE"] == 1].sort_values("TMAX", ascending=False).head(8))
```

```python
print(df["HEATWAVE"].value_counts())
```

```python
print(df["DISTRICT"].value_counts())
```

```python
print(df.loc[0:4, ["DISTRICT", "TMAX", "HEATWAVE"]])
```

```python
print(df.iloc[0:5, 0:5])
```

```python
print(df[(df["TMAX"] > 318.15) & (df["HEATWAVE"] == 1)]
      [["DISTRICT", "TMAX", "HEATWAVE"]].head())
```

```python
print(df[(df["RAIN"] == 0) & (df["TMAX"] > 318.15)]
      [["DISTRICT", "TMAX", "RAIN"]].head())
```

```python
print("Minimum TMAX:", df["TMAX"].min())
```

```python
print("Maximum TMAX:", df["TMAX"].max())
```

```python
print(df.groupby("DISTRICT")["TMAX"].mean())
```

```python
print(df.groupby(["DISTRICT", "MONTH"])["TMAX"].mean())
```

```python
df["TEMP_DIFFERENCE"] = df["TMAX"] - df["TMIN"]

print(df[["TMAX", "TMIN", "TEMP_DIFFERENCE"]].head())
```

```python
print(df.groupby("DISTRICT")["TMAX"].agg(["mean", "min", "max"]))
```

```python
print(df.groupby("DISTRICT").agg({
    "TMAX": "mean",
    "RAIN": "sum"
}))
```

```python
print(df["TMAX"].corr(df["TEMP2M"]))
```

```python
df["TMAX_NORMALIZED"] = (
    (df["TMAX"] - df["TMAX"].min()) /
    (df["TMAX"].max() - df["TMAX"].min())
)

print(df[["TMAX", "TMAX_NORMALIZED"]].head())
```

```python
group = df.groupby("DISTRICT")

print(group.get_group("Barmer"))
```

```python
a = df[["YEAR", "MONTH", "DAY", "DISTRICT", "TMAX"]].head(5)
b = df[["YEAR", "MONTH", "DAY", "DISTRICT", "HEATWAVE"]].head(5)

print("MERGE:")
print(pd.merge(a, b, on=["YEAR", "MONTH", "DAY", "DISTRICT"]))

print("\nJOIN:")
a1 = a.set_index(["YEAR", "MONTH", "DAY", "DISTRICT"])
b1 = b.set_index(["YEAR", "MONTH", "DAY", "DISTRICT"])
print(a1.join(b1))

print("\nCONCATENATION:")
c = df[df["DISTRICT"] == "Barmer"].head(3)
d = df[df["DISTRICT"] == "Jaisalmer"].head(3)
print(pd.concat([c, d]))
```
